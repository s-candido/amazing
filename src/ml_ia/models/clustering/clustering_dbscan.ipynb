### Analyse Silhouette, Calinski-Harabasz & Davies-Bouldin

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import matplotlib.pyplot as plt
import numpy as np

# Define the range of eps values to evaluate
eps_values = np.linspace(0.1, 5.0, 20)  # Adjust the range and step size as needed

# Initialize lists to store the scores
silhouette_scores = []
calinski_harabasz_scores = []
davies_bouldin_scores = []

# Evaluate clustering for each eps value
for eps in eps_values:
    dbscan = DBSCAN(eps=eps, min_samples=5)  # Adjust min_samples as needed
    labels = dbscan.fit_predict(X_scaled)
    
    # Check if there are at least 2 clusters (DBSCAN can assign -1 for noise points)
    if len(set(labels)) > 1:
        # Compute the metrics
        silhouette_scores.append(silhouette_score(X_scaled, labels))
        calinski_harabasz_scores.append(calinski_harabasz_score(X_scaled, labels))
        davies_bouldin_scores.append(davies_bouldin_score(X_scaled, labels))
    else:
        # Append NaN if clustering is not valid
        silhouette_scores.append(np.nan)
        calinski_harabasz_scores.append(np.nan)
        davies_bouldin_scores.append(np.nan)

# Plot the metrics
plt.figure(figsize=(15, 5))

# Silhouette Coefficient
plt.subplot(1, 3, 1)
plt.plot(eps_values, silhouette_scores, marker='o', linestyle='--')
plt.title("Silhouette Coefficient")
plt.xlabel("Epsilon (eps)")
plt.ylabel("Score")
plt.grid(True)

# Calinski–Harabasz Index
plt.subplot(1, 3, 2)
plt.plot(eps_values, calinski_harabasz_scores, marker='o', linestyle='--')
plt.title("Calinski–Harabasz Index")
plt.xlabel("Epsilon (eps)")
plt.ylabel("Score")
plt.grid(True)

# Davies–Bouldin Index
plt.subplot(1, 3, 3)
plt.plot(eps_values, davies_bouldin_scores, marker='o', linestyle='--')
plt.title("Davies–Bouldin Index")
plt.xlabel("Epsilon (eps)")
plt.ylabel("Score")
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Fit et Plot KMeans pour 4, 5 et 6 Clusters

```python
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
import plotly.graph_objs as go
import plotly.offline as py
from sklearn.metrics import silhouette_samples, silhouette_score
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
import pandas as pd
import numpy as np

def dbscan_fit_and_plot(eps, min_samples, X_final, df):
    df_tmp = df.copy()
    dbscan = DBSCAN(eps=eps, min_samples=min_samples, metric='euclidean', algorithm='auto', leaf_size=5)
    clusters = dbscan.fit_predict(X_final)
    df_tmp["segment"] = clusters

    # --- Silhouette Analysis ---
    if len(set(clusters)) > 1:  # Ensure there are at least 2 clusters
        print("Silhouette Analysis...")
        silhouette_avg = silhouette_score(X_final, clusters)
        print(f"For eps = {eps}, min_samples = {min_samples}, the average silhouette_score is: {silhouette_avg}")

        sample_silhouette_values = silhouette_samples(X_final, clusters)

        # Print silhouette score for each cluster
        for cluster in set(clusters):
            if cluster != -1:  # Exclude noise points
                cluster_silhouette_avg = sample_silhouette_values[clusters == cluster].mean()
                print(f"Silhouette score for Cluster {cluster}: {cluster_silhouette_avg}")
        # --- Silhouette Plot ---
        print("Generating Silhouette Plot...")
        plt.figure(figsize=(10, 8))
        y_lower = 10
        colormap = plt.colormaps.get_cmap("tab10")
        cluster_colors = [mcolors.to_hex(colormap(i)) for i in range(len(set(clusters)))]
        cluster_color_map = {cluster: cluster_colors[i] if cluster != -1 else "#000000" for i, cluster in enumerate(set(clusters))}

        for cluster in set(clusters):
            if cluster != -1:  # Exclude noise points
                ith_cluster_silhouette_values = sample_silhouette_values[clusters == cluster]
                ith_cluster_silhouette_values.sort()

                size_cluster_i = ith_cluster_silhouette_values.shape[0]
                y_upper = y_lower + size_cluster_i

                color = cluster_color_map[cluster]
                plt.fill_betweenx(
                    np.arange(y_lower, y_upper),
                    0,
                    ith_cluster_silhouette_values,
                    facecolor=color,
                    edgecolor=color,
                    alpha=0.7,
                )

                plt.text(-0.05, y_lower + 0.5 * size_cluster_i, str(cluster))
                y_lower = y_upper + 10

        plt.title(f"Silhouette analysis for DBSCAN clustering (eps = {eps}, min_samples = {min_samples})")
        plt.xlabel("The silhouette coefficient values")
        plt.ylabel("Cluster label")
        plt.axvline(x=silhouette_avg, color="red", linestyle="--")
        plt.yticks([])
        plt.xticks(np.arange(-0.1, 1.1, 0.1))
        plt.show()
    else:
        print(f"For eps = {eps}, min_samples = {min_samples}, clustering resulted in less than 2 clusters.")

    # Define consistent colors for clusters
    colormap = plt.colormaps.get_cmap("tab10")
    cluster_colors = [mcolors.to_hex(colormap(i)) for i in range(len(set(clusters)))]
    cluster_color_map = {cluster: cluster_colors[i] if cluster != -1 else "#000000" for i, cluster in enumerate(set(clusters))}  # Map cluster index to color

    # --- PCA Scatter Plot (2D) ---
    print("Visualisation du clustering (2D)...")
    pca = PCA(n_components=2)
    X_pca_2d = pca.fit_transform(X_final)

    plt.figure(figsize=(10, 8))
    for cluster in set(clusters):
        cluster_points = X_pca_2d[clusters == cluster]
        plt.scatter(cluster_points[:, 0], cluster_points[:, 1], c=cluster_color_map[cluster], label=f"Cluster {cluster}" if cluster != -1 else "Noise", s=10)

    plt.title(f"Visualisation des clusters utilisateurs (PCA 2D) - eps = {eps}, min_samples = {min_samples}")
    plt.xlabel("PCA 1")
    plt.ylabel("PCA 2")
    plt.legend()
    plt.grid(True)
    plt.show()

    # --- PCA Scatter Plot (3D) ---
    print("Visualisation du clustering (3D)...")
    pca_3d = PCA(n_components=3)
    X_pca_3d = pca_3d.fit_transform(X_final)

    # Prepare data for Plotly
    traces = []
    for cluster in set(clusters):
        cluster_points = X_pca_3d[clusters == cluster]
        trace = go.Scatter3d(
            x=cluster_points[:, 0],
            y=cluster_points[:, 1],
            z=cluster_points[:, 2],
            mode='markers',
            marker=dict(
                size=8,
                color=cluster_color_map[cluster],  # Use consistent cluster color
                opacity=0.8
            ),
            name=f"Cluster {cluster}" if cluster != -1 else "Noise"
        )
        traces.append(trace)

    layout = go.Layout(
        margin=dict(l=0, r=0, b=0, t=0),
        title=f"Interactive 3D Plot of Clusters - eps = {eps}, min_samples = {min_samples}"
    )
    fig = go.Figure(data=traces, layout=layout)
    py.iplot(fig, filename='3d-scatter-colorscale')

    return df_tmp
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_cluster_stats(df, segment_column="segment"):
    """
    Plots statistics, distributions, and boxplots for clusters in the DataFrame.

    Parameters:
    - df: DataFrame containing the clustering results and features.
    - segment_column: Name of the column containing cluster segments.

    Returns:
    - stats: A dictionary containing descriptive statistics for each cluster.
    """
    # Ensure the segment column exists
    if segment_column not in df.columns:
        raise ValueError(f"Column '{segment_column}' not found in the DataFrame.")

    # Number of individuals per cluster
    cluster_counts = df[segment_column].value_counts().sort_index()
    print("Number of individuals per cluster:")
    print(cluster_counts)

    # Plot the distribution of individuals per cluster
    plt.figure(figsize=(10, 6))
    cluster_counts.plot(kind="bar", color="skyblue", alpha=0.8)
    plt.title("Distribution of Individuals per Cluster")
    plt.xlabel("Cluster")
    plt.ylabel("Number of Individuals")
    plt.xticks(rotation=0)
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    plt.show()

    # Descriptive statistics for each cluster
    stats = {}
    for cluster in cluster_counts.index:
        cluster_data = df[df[segment_column] == cluster]
        cluster_stats = cluster_data.describe()
        stats[cluster] = cluster_stats
        print(f"\nDescriptive statistics for Cluster {cluster}:")
        print(cluster_stats)

    # Boxplots for each column grouped by cluster
    numeric_columns = df.select_dtypes(include=["number"]).columns
    numeric_columns = [col for col in numeric_columns if col != segment_column]

    for col in numeric_columns:
        plt.figure(figsize=(12, 6))
        sns.boxplot(x=segment_column, y=col, data=df, palette="Set3")
        plt.title(f"Boxplot of {col} by Cluster")
        plt.xlabel("Cluster")
        plt.ylabel(col)
        plt.grid(axis="y", linestyle="--", alpha=0.7)
        plt.show()

    return stats
```
